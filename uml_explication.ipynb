## On a décidé de monter notre DB sur la base du modèle en étoile (star schema) : sépare notre db en deux types de tables : les dimensions(dim) et les Faits (facts)

## 1. Les tables de Dimensions (Le "Contexte")
il s'agit des axes d'analyse. Elles répondent aux questions Qui ? Quoi ? Où ? Quand ?

#### dim_hebergement : Le "Quoi". Contient toutes les infos descriptives d'un hôtel qui changent peu (nom, classification, capacité).

dim_localisation : Le "Où". Gère la géographie. Séparer cela va permettre  de filtrer facilement "tous les hôtels d'une région donnée" sans toucher aux autres données.

#### dim_date et dim_temps  : Le "Quand".  Au lieu de manipuler des formats de date complexes en SQL, on va utiliser des "Smart Keys" (ex: 20260405). Ça va  permettre aux outils comme POWERBI/Metabase de faire des tris par mois, trimestre ou année plus rapidemment .

#### dim_equipement : Le dictionnaire unique de tous les équipements possibles (Piscine, WiFi, etc.).

## 2. Les tables de Faits (Le "Cœur" de l'étoile)
Elles vont relier toutes les tables dimensions entre elles.

#### fact_hebergement : La table principale. Elle ne contiendra  presque que des identifiants (clés étrangères) et des métriques. Elle dit : "Tel jour, à telle heure, tel hôtel situé à tel endroit avait telle classification".

#### fact_equipement_hebergement : il s'agira d'une  table de liaison (Bridge Table). Puisqu'un hôtel a plusieurs équipements, et qu'un équipement appartient à plusieurs hôtels (relation Many-to-Many), cette table sert de pont entre dim_hebergement et dim_equipement.
